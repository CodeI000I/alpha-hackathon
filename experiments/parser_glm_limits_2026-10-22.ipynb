## Код live-прогона

Ячейка ниже показывает выполненный код. Её повторное выполнение отправляет до трёх новых платных запросов с локальным ключом. Python 3.13.15; запускать из корня репозитория. Сохранённый output получен из фактического results.json. Проверки не меняют ответ модели.

```python
import asyncio, json, logging, time
from pathlib import Path
from pydantic import SecretStr
from parser_service.extraction_models import SourceDocument
from parser_service.ai_extract import extract_document, ExtractionError
from parser_service.llm_client import open_extraction_model, MODEL_ID

OUT = Path('/private/tmp/kib-parser-live-high-grounding')
OUT.mkdir(parents=True, exist_ok=True)
OUT.mkdir(parents=True, exist_ok=True)
ROOT = Path.cwd()
if ROOT.name == "experiments":
    ROOT = ROOT.parent
for name in ('openai', 'httpx', 'httpcore', 'langsmith'):
    logging.getLogger(name).disabled = True

def read_key():
    lines = (ROOT / '.local-context/.env.llm').read_text().splitlines()
    for line in lines:
        name, separator, value = line.partition('=')
        if separator and name.strip() == 'HACKATHON_LLM_API_KEY':
            value = value.strip()
            if len(value) >= 2 and value[0] == value[-1] and value[0] in ('\"', "'"):
                value = value[1:-1]
            return SecretStr(value)
    raise RuntimeError('Local gateway key is missing.')

async def main():
    documents = snapshots
    results = []
    async with open_extraction_model(api_key=read_key(), max_output_tokens=16384, timeout_seconds=180) as model:
        wire = []
        async def record(response):
            await response.aread()
            item = {'http_status': response.status_code}
            if response.is_success:
                body = response.json()
                item.update(usage=body.get('usage'), response_model=body.get('model'))
                choice = (body.get('choices') or [{}])[0]
                item['finish_reason'] = choice.get('finish_reason')
                item['raw_content'] = (choice.get('message') or {}).get('content')
            wire.append(item)
        model.http_async_client.event_hooks['response'].append(record)
        for row in documents[:3]:
            entry = {k: row[k] for k in ('index','url','rss_title','rss_publication')}
            if 'document' not in row:
                entry.update(status='source_failed', error=row.get('error'))
                results.append(entry)
                continue
            print(json.dumps({'index':row['index'], 'stage':'model_request_started'}, ensure_ascii=False), flush=True)
            source = SourceDocument.model_validate(row['document'])
            started = time.monotonic()
            before = len(wire)
            try:
                result = await extract_document(source, model, max_input_chars=60000, max_output_chars=60000, timeout_seconds=180)
                entry.update(status='validated', result=result.model_dump(mode='json'))
            except ExtractionError as e:
                entry.update(status='extraction_failed', error=e.code)
            entry.update(reasoning_effort='high', max_output_tokens=16384, elapsed_seconds=round(time.monotonic()-started,2), model=MODEL_ID, text_hash=source.text_hash, loader_incomplete_fields=row['incomplete_fields'])
            if len(wire) > before:
                entry['response'] = wire[-1]
            results.append(entry)
            (OUT/'results.json').write_text(json.dumps(results,ensure_ascii=False,indent=2))
            print(json.dumps({k:v for k,v in entry.items() if k not in ('result','response','text_hash')}, ensure_ascii=False), flush=True)
            if row['index'] == 1 and entry['status'] != 'validated':
                print('Stopped after failed canary; no further model requests.', flush=True)
                break
            if entry.get('response',{}).get('http_status') in (401,403,402,429):
                print('Stopped after gateway access/budget/rate failure; no retries.', flush=True)
                break
    (OUT/'results.json').write_text(json.dumps(results, ensure_ascii=False, indent=2))

await main()
```

Независимое смысловое ревью: первая новость без существенных замечаний; во второй остаются event_date, выведенные из публикационной даты; в третьей неверно назван индикатор среднего максимума, а цитаты не покрывают все детали составных claims. Сырые ответы не исправлялись. Эти поля нельзя использовать как подтвержденные. Краткие пересказы, проверенные основной моделью по источникам, записаны отдельно в semantic_review.human_checked_summaries.

Невалидные ответы сохранены отдельно и не исправлялись после генерации. Ошибочные и промежуточные ответы выше не являются принятыми результатами. Валидность цитат доказывает их наличие в источнике; она не заменяет смысловую оценку каждого утверждения. Полный HTTP jobs → БД → Go/LangGraph в этом эксперименте не запускался. Блок пользователем не закрыт.
