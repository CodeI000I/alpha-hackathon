## Фактические ответы

Сетевая ячейка ниже содержит код выполненного эксперимента. Её повторный запуск делает ещё до трёх платных запросов с локальным ключом. Код запускался основным агентом в Python 3.13.15 из корня проекта; сохранённые outputs получены из фактического `results.json`. Заголовки и ключ не сохранялись. Чтение HTTP-ответа здесь служит только диагностике: проверка JSON/цитат выполняется неизменённым `extract_document`.

```python
import asyncio, json, logging, time
from pathlib import Path
from pydantic import SecretStr
from parser_service.extraction_models import SourceDocument
from parser_service.ai_extract import extract_document, ExtractionError
from parser_service.llm_client import open_extraction_model, MODEL_ID

OUT = Path('/private/tmp/kib-parser-live-three')
OUT.mkdir(parents=True, exist_ok=True)
ROOT = Path.cwd()
if ROOT.name == "experiments":
    ROOT = ROOT.parent
for name in ('openai', 'httpx', 'httpcore', 'langsmith'):
    logging.getLogger(name).disabled = True

def read_key():
    lines = (ROOT / '.local-context/.env.llm').read_text().splitlines()
    for line in lines:
        name, separator, value = line.partition('=')
        if separator and name.strip() == 'HACKATHON_LLM_API_KEY':
            value = value.strip()
            if len(value) >= 2 and value[0] == value[-1] and value[0] in ('\"', "'"):
                value = value[1:-1]
            return SecretStr(value)
    raise RuntimeError('Local gateway key is missing.')

async def main():
    documents = snapshots
    results = []
    async with open_extraction_model(api_key=read_key(), max_output_tokens=4096, timeout_seconds=55) as model:
        wire = []
        async def record(response):
            await response.aread()
            item = {'http_status': response.status_code}
            if response.is_success:
                body = response.json()
                item.update(usage=body.get('usage'), response_model=body.get('model'))
                choice = (body.get('choices') or [{}])[0]
                item['finish_reason'] = choice.get('finish_reason')
                item['raw_content'] = (choice.get('message') or {}).get('content')
            wire.append(item)
        model.http_async_client.event_hooks['response'].append(record)
        for row in documents[:3]:
            entry = {k: row[k] for k in ('index','url','rss_title','rss_publication')}
            if 'document' not in row:
                entry.update(status='source_failed', error=row.get('error'))
                results.append(entry)
                continue
            print(json.dumps({'index':row['index'], 'stage':'model_request_started'}, ensure_ascii=False), flush=True)
            source = SourceDocument.model_validate(row['document'])
            started = time.monotonic()
            before = len(wire)
            try:
                result = await extract_document(source, model, max_input_chars=60000, max_output_chars=24000, timeout_seconds=60)
                entry.update(status='validated', result=result.model_dump(mode='json'))
            except ExtractionError as e:
                entry.update(status='extraction_failed', error=e.code)
            entry.update(elapsed_seconds=round(time.monotonic()-started,2), model=MODEL_ID, text_hash=source.text_hash, loader_incomplete_fields=row['incomplete_fields'])
            if len(wire) > before:
                entry['response'] = wire[-1]
            results.append(entry)
            (OUT/'results.json').write_text(json.dumps(results,ensure_ascii=False,indent=2))
            print(json.dumps({k:v for k,v in entry.items() if k not in ('result','response','text_hash')}, ensure_ascii=False), flush=True)
            if entry.get('response',{}).get('http_status') in (401,403,402,429):
                print('Stopped after gateway access/budget/rate failure; no retries.', flush=True)
                break
    (OUT/'results.json').write_text(json.dumps(results, ensure_ascii=False, indent=2))

await main()
```
